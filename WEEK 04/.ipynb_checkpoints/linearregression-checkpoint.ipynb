```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python
a = np.array([[1],[2],[3],[4],[5]])#data loading(a is hours spent studying, w is exsm score)
w = np.array([3, 4, 5, 6, 7])
a_train, a_test, w_train, w_test = train_test_split(a, w, test_size=0.2)#preprocessing
model = LinearRegression()#model training
model.fit(a_train, w_train)
prediction = model.predict([[6]])
print(prediction)
```
