```python
import sqlite3
conn = sqlite3.connect("jaydee.db") #TASK 3
cursor = conn.cursor()
```

```python
cursor.execute("SELECT * FROM students WHERE age >= 20;") #DISPLAYING STUDENTS ABOVE OR EQUAL TO 20 YEARS
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("SELECT * FROM students ORDER BY age DESC;") #ARRANGING STUDENTS IN DESCENDING ORDER
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("SELECT age, COUNT(*) FROM students GROUP BY age")# WE ARE GROUPING STUDENTS BY AGE
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("SELECT * FROM students LIMIT 5;")# SHOWS ONLY THE FIRST FIVE ROWS
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("SELECT AVG(age) FROM students")#SHOWS THE AVERAGE AGE OF THE STUDENTS
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
conn.commit()
```

```python
conn.close()
```
