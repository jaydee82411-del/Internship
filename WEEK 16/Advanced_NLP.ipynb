```python
#TASK 2 USIBG PRETTRAINED WORD EMBEDDINGS
#importing libraries
import gensim.downloader as api
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
#load the pre trained GloVe word vectors
glove = api.load("glove-wiki-gigaword-50")
```

```python
glove["king"]
```

```python
glove.most_similar("king")
```

```python
#USING THE EMBEDDINGS FOR CLASSIFICATION
data = [
    "i love the book",
    "this is a great book",
    "the fit is great",
    "i love the shoes",
    "i hate this book"
]

sentiment = [
    1,1,1,1,0
]
#1 = positive
#0 = negative
```

```python
#convert sentences into ebeddings
def sentence_vector(sentence):
  words = sentence.lower().split()

  vectors = [glove[word] for word in words if word in glove]
  return np.mean(vectors, axis=0)
```

```python
#now, lets create the target and features
x = np.array([sentence_vector(text) for text in data])
y = np.array(sentiment)
```

```python
x.shape
```

```python
#TRAIN A SIMPLE CLASSIFIER (logistic regression specifically)
#everything before here was like one hot encoding, but a smarter one in the sense
#it can recognise words and even relationship between words

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=45) #Data splitting

model = LogisticRegression() #model loading

model.fit(x_train, y_train) #model training
```

```python
#predictions
y_pred = model.predict(x_test)
y_pred
```

```python
#evaluation
accuracy = accuracy_score(y_test, y_pred)
accuracy
```
