```python
#import libraries
import tensorflow as tf
import numpy as np
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM
from sklearn.metrics import mean_squared_error
```

```python
#this is going to be my sequence data (samples, time steps, features)(I'm looking at it like features
#from linear regression)
X = np.array([
    [[1], [2], [3]],
    [[2], [4], [6]],
    [[3], [4], [5]],
    [[4], [8], [16]]
], dtype=np.float32)

#(looking at y as target)
y = np.array([4, 8, 6, 32])
```

```python
#creating the neural network (model creation from linear regression)
model = Sequential([
    LSTM(50, input_shape=(3, 1)),
    Dense(1)
])
```

```python
#configure hoe the model works (i think likw hyper parameter tuning)
model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)
```

```python
#training the model
model.fit(
    X,y,epochs=100,verbose=1
)
```

```python
#we are creating a new array to test the model but we're not using y cause we didnt split data into traing and test
new_data = np.array([
    [[5], [10], [15]]
], dtype=np.float32)

#now, the prediction
pred = model.predict(new_data)
pred
```

```python
#evaluation
mse = mean_squared_error(y, model.predict(X))
mse
```
