```python
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()#data loading
a, b = iris.data, iris.target
a_train, a_test, b_train, b_test = train_test_split(a, b, test_size=0.3, random_state=41)#preprocessing
knn = KNeighborsClassifier(n_neighbors=4)#model training
knn.fit(a_train, b_train)
prediction = knn.predict(a_test)#evaluation 
accuracy = accuracy_score(b_test, prediction)
print(accuracy)
```
