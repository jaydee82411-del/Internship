```python
# importing libraries
import nltk
import spacy
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.naive_bayes import MultinomialNB

nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("punkt_tab")
```

```python
#first lets get a simple sentece we can use to practice
text = "i really enjoyed learning Natural Language Processing"

#Tokenization
tokens = word_tokenize(text)
tokens
```

```python
stop_words = set(stopwords.words("english"))
filtered_words = [word for word in tokens if word.lower() not in stop_words]
filtered_words
```

```python
#stemming
stemmer = PorterStemmer()
stemmed_words = [
    stemmer.stem(word)
    for word in filtered_words]
stemmed_words
```

```python
#lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_words
]
lemmatized_words
```

```python
#simple text classification model
data = {
    "text": [
        "i love this product",
        "this product is amazing",
        "this is fantastic",
        "i am very happy with this",
        "the products work perfectly",
        "i hate this product",
        "this product is terrible",
        "i really dislike it",
        "this is awful",
        "i amm disappointed",
        "the product does not work"
    ],
    "sentiment": [
        "positive",
        "positive",
        "positive",
        "positive",
        "positive",
        "negative",
        "negative",
        "negative",
        "negative",
        "negative",
        "negative"

    ]
}

df = pd.DataFrame(data)
df
```

```python
#TF-IDF features, (like feature scaling in linear regression)
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(df["text"])
y = df["sentiment"]

#now split the data into training and testing set
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=45, stratify=y)

#training the model
model = MultinomialNB()
model.fit(X_train, y_train)

#making predictions
y_pred = model.predict(X_test)
y_pred
```

```python
#model evaluation
accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
#SVM model
svm_model = SVC() #model loading
svm_model.fit(X_train, y_train) #model training

svm_pred = svm_model.predict(X_test) #prediction

svm_accuracy = accuracy_score(y_test, svm_pred)#evaluation
svm_accuracy
```

```python
#SENTIMENT ANALYSIS ON A SMALL DATASET
new_data = [
    "i hate this product",
    "this product is amazing",
    "this is fantastic",
    "i am very happy with this",
    "the products work perfectly"
]

#scaling
new_data_features = vectorizer.transform(new_data)

#predictions
predictions = model.predict(new_data_features)
predictions
```
