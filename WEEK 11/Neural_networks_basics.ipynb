```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np
```

```python
x = np.array([[1], [2], [3], [4]])
y = np.array([[3], [6], [9], [12]])
```

```python
model = Sequential([
    Dense(10, activation='relu', input_shape=[1]),
    Dense(1)
])
```

```python
model.compile(
    optimizer='adam',
    loss='mse'
)
```

```python
model.fit(x, y, epochs=500)
```

```python
prediction = model.predict(np.array([[9]]))
prediction
```
