```python
#importing libraries
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
#lload the dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
```

```python
#normalize the images(model trains better when the value is smaller(like feature scaling, just that we are convering pixel values))
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
#now we build the model
model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3, 3), activation="relu", padding="same",
                           input_shape=(32,32,3)),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])
```

```python
#model copiling (in a way it's like hyoer oarameter tuning but instead of tweaking, we are chooosing settings)
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
#look at the model
model.summary()
```

```python
#model training
history = model.fit(
    x_train,
    y_train,
    epochs=30,
    validation_split=0.1#btw, validation split works exactly like test_size, sets aside a percentage of the data for testing
)
```

```python
#prediction
predictions = model.predict(x_test[:1])
print(tf.argmax(predictions[0]))
```

```python
#classes for the groups according to the dataset are
classes = ["airplane", "automobile", "bird", "cat", "deer", "dog", "frog", "horse", "ship", "truck"]
```

```python
#prediction v actual result
print("predicted:",
      classes[tf.argmax(predictions[0])]
) #predicted
print("actual:",
      classes[y_test[0][0]]
) #actual
```

```python
#model evaluation
accuracy = model.evaluate(x_test, y_test)
accuracy
```

```python
#TASK 4
#EXPERIMENT WITH DIFFERENT CNN ACHITECTURES
#We'll make the model deeper so..
model2 = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3, 3), activation="relu", padding="same",
                           input_shape=(32,32,3)),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(128, (3, 3), activation="relu", padding="same"),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])
```

```python
#compile 2nd model
model2.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
#train it
history2 = model2.fit(
    x_train,
    y_train,
    epochs=5,# I waited a long time the last time that is why i am reducing it :)
    validation_split=0.1
)
```

```python
#now we evakuate
test_accuracy = model2.evaluate(x_test, y_test)
test_accuracy
```
