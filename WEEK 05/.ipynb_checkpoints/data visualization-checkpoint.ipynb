```python
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset("titanic") #data loading
print(df.head())
```

```python
print(df.tail())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.histplot(df["age"].dropna(), bins=30, kde=True, color="green")#histogram for distribution of passenger ages
plt.title("Ddistribution of passenger ages")
plt.xlabel("age")
plt.ylabel("frequency")
plt.show
```

```python
sns.barplot(x="sex", y="survived", data=df, palette="viridis")#barchart for survival rate by gender
plt.title("Survival rate by gender")
plt.show()
```

```python
sns.scatterplot(x="age", y="fare", hue="survived", data=df)#don't really understand how scatterplot works, i'd read further
plt.title("Age vs Fare (colored by survival)")
plt.show()
          
```

```python
sns.boxplot(x="pclass", y="age", data=df)#box plot is also not immediately comprehensive
plt.title("Age distribution by passenger class")
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation heatmap of titanic features")
plt.show()
            
```
