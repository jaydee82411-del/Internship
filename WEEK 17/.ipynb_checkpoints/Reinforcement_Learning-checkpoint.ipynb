```python
#TASK 2
#Understanding RL Enviroment
#importing libraries
import gymnasium as gym
```

```python
#Creating the enviroment
env = gym.make("FrozenLake-v1", is_slippery=False)
```

```python
state, info = env.reset()
state #we're resetting the initial state to 0
```

```python
print("Number of states:", env.observation_space.n)
print("number of actions:", env.action_space.n) #no of states and possible actons, i.e, 4 actions, i.e up, down, left...
```

```python
#TASK 3, BUILDING THE Q-LEARNING ALGORITHM
#importing libraries
import numpy as np
```

```python
#create q-table
q_table = np.zeros(
    (env.observation_space.n, env.action_space.n)
)
q_table
```

```python
#learning parameters
learning_rate = 0.8
discount_factor = 0.95
episodes = 1000

#exploration parameters
epsilon = 1.0
epsilon_decay = 0.995
epsilon_min = 0.01
```

```python
#q_learning agent
for episode in range(episodes):
  start, info = env.reset()
  done = False

  while not done:
    if np.random.random() < epsilon: #this generates a random number between 1 and 0 and tells the agent to explore if its less than epsilon
      action = env.action_space.sample()#choose a random action
    else:
      action = np.argmax(q_table[state])#choose the action with highest q_value

    next_state, reward, terminated, truncated, info = env.step(action)

    done = terminated or truncated#episode done if its truncated or terminated

    q_table[state, action] += learning_rate * (
        reward + discount_factor * np.max(q_table[next_state]) - q_table[state, action]
    ) #update the q_table using q_learning

    state = next_state#move the agent to new state after updating q_table and start next loop

  epsilon *= epsilon_decay #reduce esploartion after each episode so that the agent relies more on what it has learnt

```

```python
q_table
```

```python
#visualize learned policy
direction = ["L", "D", "R", "U"]
policy = np.argmax(q_table, axis=1)

for row in range(4):
  print("".join(direction[policy[row * 4 + col]] for col in range(4)))
```
